# CODE: Entropy 

```python
import numpy as np 
```

```python
Y = np.array([1,0,0,1,0,1,0,1])
```

```python
def entropy(var): 
    N = var.shape[0]
    values, count = np.unique(var, return_counts=True)
    
    ent = 0.0 

    for i in count: 
        p = i/N 
        ent += (p * np.log2(p)) 
    return -ent 
```

```python
entropy(Y)
```

```python
entropy(np.array([1,1,1,1,1,1,1]))
```

## CODE: Split Data

```python
import pandas as pd 
```

```python
df = pd.read_csv('golf.csv')
```

```python
df
```

```python
def divide_data(data, feature): 
    DATA = {}
    fea_values = list(df[feature].value_counts().index)
    occurence = list(df[feature].value_counts())

    for val in fea_values: 
        # Create empty dictionary which has data(empty DataFrame) and len = 0 initially
        DATA[val] = {'data' : pd.DataFrame([], columns = data.columns), 'len': 0}

    for ix in range(data.shape[0]): 
        val = data[feature].iloc[ix]

        # DATA[val]['data'] = DATA[val]['data'].append(data.iloc[ix])
        DATA[val]['data'] = pd.concat([DATA[val]['data'], data.iloc[[ix]]])

        idx = fea_values.index(val) 
        DATA[val]['len'] = occurence[idx]

    return DATA 
```

```python
divide_data(df, 'Outlook')
```

```python
divide_data(df, 'Windy')
```

# Information Gain

```python
def information_gain(data, feature): 
    examples = data.shape[0]

    DATA = divide_data(data, feature) 
    keys = DATA.keys()

    entropy_of_children = 0.0 
    for key in keys: 
        entropy_of_children += ( (DATA[key]['len']/examples)*entropy(DATA[key]['data']['Play']) )

    IG = entropy(data['Play']) - entropy_of_children
    return IG
```

```python
information_gain(df, 'Outlook')
```

```python
information_gain(df, 'Windy')
```

```python
information_gain(df, 'Temperature')
```

```python
information_gain(df, 'Humidity')
```
