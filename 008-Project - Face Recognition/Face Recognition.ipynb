# Object Detection 
- ##### Haar Cascades Classifier -> Less Memory, Fast, Less Accurate- ##### Modern Deep Learning -> More Memory, Slow, More Accurate


## Haar Cascades Classifier

##### Haar cascade is an classifier that can detect object in images, irrespective of their scale in image and location. We can train a haar-cascade detector to detect various objects like cars, bikes, buildings, fruits, etc  

- Haar cascades are fast and can work well in real-time
- Haar cascades is not as accurate as modern object detection techniques are.
- Haar cascades has a downside. It predicts many false positive
- Simple to implement, less computing power required.

```python
img = cv2.imread('groupPhoto.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
model = cv2.CascadeClassifier("haarcascade_frontalface_alt.xml")
```

```python
faces = model.detectMultiScale(img, 1.3, 5)
```

```python
faces.shape
```

```python
#model.detectMultiScale?
```

```python
for f in faces:
    x, y, w, h = f
    cv2.rectangle(img, (x,y),(x+w,y+h),(0,255,0),3)
plt.imshow(img)
plt.show()
```
