# Feature Selection  

- Feature selection is the process of selecting a subset of relevant features from a larger set of features in a dataset. The goal is to identify the most informative and discriminative features that contribute significantly to the predictive power of the model. By selecting the right set of features, we can improve the model’s accuracy, reduce overfitting, and enhance interpretability.

# Dimensionality reduction   

- Dimensionality reduction refers to techniques that transform a high-dimensional dataset into a lower-dimensional representation while preserving its essential structure and characteristics. The aim is to reduce the computational complexity, improve visualization, and eliminate redundant or noisy features.

### Let's drive into code 

```python
import numpy as np
import pandas as pd 
```

```python
data = pd.read_csv('train.csv')
data.shape
```

```python
data.head()
```

```python
X = data.iloc[:, :-1]
y = data.iloc[:, -1]
```

```python
print(X.shape, y.shape)
```

```python
#Normalise/Standardise data 
mu = X.mean(axis=0)
sig = X.std(axis=0)

X = (X-mu)/sig 
```

```python
from sklearn.feature_selection import SelectKBest
```

```python
best_features = SelectKBest()
```

```python
fit = best_features.fit(X, y) 
```

```python
fit.scores_
```

```python
dfcolumns = pd.DataFrame(X.columns)
dfscores = pd.DataFrame(fit.scores_)
```

```python
featureScore = pd.concat([dfcolumns,dfscores], axis=1)
```

```python
featureScore.columns = ['Features', 'Scores']
```

```python
featureScore
```

```python
featureScore = featureScore.sort_values(by='Scores', ascending=False)
```

```python
featureScore
```

```python
featureScore[:10]['Features'].values   # Top 10 features
```

```python
top_10 = list(featureScore[:10]['Features'].values)
top_10
```

# Model Comparison 

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
```

```python
model = LogisticRegression()
model.fit(X,y)
```

```python
# For all 20 features of X
scores = cross_val_score(model, X, y, cv=10)
scores.mean()
```

```python
# For only top 10 features of X
scores = cross_val_score(model, X[top_10], y, cv=10)
scores.mean()
```
