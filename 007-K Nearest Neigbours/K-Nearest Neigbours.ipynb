```python
from sklearn.datasets import make_blobs
```

```python
#KNN Data Prep
X, y = make_blobs(n_samples=2000, n_features=2, cluster_std=3, centers=3, random_state=42)
n_features = X.shape[1]
m = X.shape[0] 
```

```python
print(X.shape, y.shape)
```

```python
for i in range(m):
    if y[i]==0: 
        plt.scatter(X[i,0], X[i,1], c='r', label="red")
    elif y[i]==1: 
         plt.scatter(X[i,0], X[i,1], c='g', label="green")
    else: 
         plt.scatter(X[i,0], X[i,1], c='b', label="blue")

Xt = np.array([-5, 1])
plt.scatter(Xt[0], Xt[1], color='orange', marker='*')
plt.show()
```

```python
# KNN 
```

```python
def dist(p,q): 
    return np.sqrt(np.sum((p-q)**2))

def knn(X, y, Xt, k=5): 
    dlist = [] 
    for i in range(m): 
        d = dist(X[i], Xt) 
        dlist.append((d,y[i])) # append distance and label using tuple
    
    dlist = sorted(dlist)
    #print(dlist[:k])
    dlist = np.array(dlist[:k])
    labels = dlist[:,1]
    print(labels)

    labels, cnts = np.unique(labels, return_counts=True)
    print(labels, cnts)

    maxIdx = cnts.argmax()
    pred = labels[maxIdx]

    return int(pred) 
```

```python
knn(X,y,Xt, k=6)
```
